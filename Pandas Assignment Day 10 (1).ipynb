## Groupby

```python
# Create a dataframe using dictionary of your choice

data ={"Name":["ajay","neha","aman","ekta","ajay","neha"],
       "Age":[40,34,29,38,40,34],
       "Salary":[60000,45000,30000,50000,70000,55000]}
df=pd.DataFrame(data)
print(df)
       
```

```python
# Use Groupby of single column with aggregate sum()

df.groupby('Name')['Salary'].sum()
```

```python
# Use Groupby of single column with aggregate count()

df.groupby('Name')['Salary'].count()
```

```python
# Use Groupby of single column with aggregate min() and max()

print(df.groupby('Name')['Salary'].min())
print(df.groupby('Name')['Salary'].max())
```

```python
# Use Groupby of any 2 columns with aggregate mean()

df.groupby(['Name','Age'])['Salary'].mean()
```

```python
# Use Groupby of any 2 columns with aggregate min() and max()

df.groupby(['Name','Age'])['Salary'].min()
df.groupby(['Name','Age'])['Salary'].max()
```
